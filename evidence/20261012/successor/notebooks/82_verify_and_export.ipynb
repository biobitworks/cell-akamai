**Sharing caveat:** outputs are limited to workshop prompts and synthetic ECA items, but raw Prometheus may reveal private operational metadata. Inspect `runs/` before sharing. Set `INCLUDE_RAW=False` to omit raw Prometheus and model output text (this prevents full independent recomputation).

```python
import sys,os,json
from pathlib import Path
P=Path.cwd().resolve(); PACKAGE=next((p for p in (P,*P.parents) if (p/'cell_akamai_handoff.py').is_file()),None)
assert PACKAGE;sys.path.insert(0,str(PACKAGE))
import cell_akamai_handoff as cak
print('Run folders:')
for p in sorted(cak.RUNS.iterdir()):
    if p.is_dir(): print(p.name, ' / ', ','.join(x.name for x in p.glob('*.json')))
```

```python
# Recheck all ECA score receipts before packaging.
for p in sorted(cak.RUNS.glob('M09_ECA_B0_B3_*')):
    if (p/'analysis.json').is_file():
        result=cak.score_eca(p)
        print('RECHECK PASS:',p.name,'n=',result['sample_size'])
    else:
        print('INCOMPLETE (no score yet):',p.name)
INCLUDE_RAW=True
RESULT_ZIP=cak.export_zip(include_raw=INCLUDE_RAW)
print('Export:',RESULT_ZIP,'Bytes:',RESULT_ZIP.stat().st_size)
print('SHA256:',cak.digest_bytes(RESULT_ZIP.read_bytes()))
```
